# Cafe Sales Analysis – Final Report

## 1. Introduction

#The Cafe Sales Analysis project aims to analyze cafe transaction data to understand product performance, customer purchasing patterns, revenue trends, payment preferences, and sales channels. Using Python, Pandas, and data visualization libraries, this project identifies meaningful business insights to support data-driven decision-making.


## 2. Dataset Overview

The dataset contains 10,000 cafe transaction records and 8 columns. It represents customer purchases, product details, payment methods, sales locations, and transaction dates.

### Dataset Columns

| Column | Description |
|---|---|
| Transaction ID | Unique transaction identifier |
| Item | Product purchased |
| Quantity | Number of units purchased |
| Price Per Unit | Price of one unit |
| Total Spent | Total transaction amount |
| Payment Method | Payment method used |
| Location | In-store or takeaway |
| Transaction Date | Date of transaction |

### Dataset Period

- Start Date: 01 January 2023
- End Date: 31 December 2023
- Total Records: 10,000
- Total Columns: 8



## 3. Data Cleaning and Preprocessing

Data cleaning was performed to improve data quality and ensure reliable analysis.

### Cleaning Steps

1. Checked missing values in all columns.
2. Identified and handled missing and invalid transaction dates.
3. Standardized inconsistent product names (`Unknow` to `Unknown`).
4. Checked numeric columns for correct data types.
5. Verified transaction totals using Quantity × Price Per Unit.
6. Checked for duplicate rows and duplicate transaction IDs.
7. Preserved Unknown categories instead of assuming their original values.

### Data Quality Findings

- Total records: 10,000
- Missing transaction dates: 460
- Unknown products: 969 (9.69%)
- Unknown payment methods: 3,178 (31.78%)
- Unknown locations: 3,961 (39.61%)
- Duplicate rows: 0
- Duplicate transaction IDs: 0
- Numeric missing values: 0
- Transaction total mismatches: 0



## 4. Exploratory Data Analysis (EDA)

Exploratory Data Analysis was performed to identify patterns, trends, and relationships in cafe sales data.

### 4.1 Product Performance

- Coffee recorded the highest quantity sold (3,551 units).
- Salad generated the highest known-product revenue (16,959).
- Salad had higher revenue per unit (approximately 4.89) than Coffee (approximately 2.04).
- Cookie recorded the lowest known-product revenue (3,651).

### 4.2 Monthly Sales Performance

- June recorded the highest monthly revenue (7,417.5).
- February recorded the lowest monthly revenue (6,673).
- October recorded the highest number of transactions (838).
- February recorded the lowest number of transactions (727).

### 4.3 Customer Payment Preferences

- Digital Wallet was the most frequently recorded known payment method (2,291 transactions).
- Credit Card recorded 2,273 transactions.
- Cash recorded 2,258 transactions.
- 3,178 transactions had an Unknown payment method.

### 4.4 Sales Location Analysis

- In-store revenue: 27,223.
- Takeaway revenue: 26,686.5.
- Unknown location revenue: 35,572.

### 4.5 Transaction Analysis

- Average transaction value: approximately 8.95.
- Juice recorded the highest number of transactions among known products (1,171).
- Coffee recorded the highest quantity sold.
- Salad generated the highest known-product revenue.

### 4.6 Product Quantity and Revenue Relationship

The product comparison showed that higher sales quantity does not necessarily result in higher revenue. Salad generated more revenue than Coffee despite having a slightly lower sales quantity.

## 6. Key Business Insights and Recommendations

Based on the exploratory data analysis, the following business insights and recommendations were identified.

### 6.1 Product Performance
**Insight:** Coffee has the highest sales quantity, while Salad generates the highest revenue.

**Recommendation:** Maintain Coffee's sales momentum and explore opportunities to promote Salad because of its higher revenue per unit.

### 6.2 Low-Revenue Products
**Insight:** Cookie has the lowest known-product revenue, despite recording 3,249 units sold.

**Recommendation:** Avoid discontinuing Cookie immediately. Evaluate its production costs and test a small price adjustment before making a decision.

### 6.3 Payment Methods
**Insight:** Digital Wallet is the most frequently recorded known payment method. However, 31.78% of payment methods are Unknown.

**Recommendation:** Improve payment-method recording at the billing counter and maintain reliable digital payment facilities.

### 6.4 Sales Locations
**Insight:** In-store revenue is slightly higher than Takeaway revenue, while 39.61% of location records are Unknown.

**Recommendation:** Improve location data collection and evaluate opportunities to improve the in-store customer experience.

### 6.5 Monthly Sales Trends
**Insight:** June recorded the highest revenue, while October recorded the highest transaction count.

**Recommendation:** Investigate monthly product-level sales and seasonal factors to understand changes in customer purchasing patterns.



## 7. Project Limitations
